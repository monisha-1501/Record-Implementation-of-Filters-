```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("Dog.jpg")
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
averaging = cv2.blur(image_rgb, (5,5))

plt.imshow(averaging)
plt.title("Averaging Filter")
plt.axis("off")
plt.show()
```

```python
kernel = np.array([[1,2,1],
                   [2,4,2],
                   [1,2,1]], dtype=np.float32) / 16

weighted = cv2.filter2D(image_rgb, -1, kernel)

plt.imshow(weighted)
plt.title("Weighted Averaging Filter")
plt.axis("off")
plt.show()
```

```python
gaussian = cv2.GaussianBlur(image_rgb, (5,5), 0)

plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
median = cv2.medianBlur(image_rgb, 5)

plt.imshow(median)
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
kernel = np.array([[0,-1,0],
                   [-1,5,-1],
                   [0,-1,0]])

sharpen = cv2.filter2D(image_rgb, -1, kernel)

plt.imshow(sharpen)
plt.title("Laplacian Sharpening")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

laplacian = cv2.Laplacian(gray, cv2.CV_64F)

laplacian = cv2.convertScaleAbs(laplacian)

plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Operator")
plt.axis("off")
plt.show()
```

```python
titles = ["Original Image", "Averaging Filter",
          "Weighted Averaging", "Gaussian Filter",
          "Median Filter", "Laplacian Sharpening",
          "Laplacian Operator"]

images = [image_rgb, averaging, weighted,
          gaussian, median, sharpen, laplacian]

plt.figure(figsize=(15,10))

for i in range(7):
    plt.subplot(2,4,i+1)
    plt.imshow(images[i], cmap="gray" if i==6 else None)
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
